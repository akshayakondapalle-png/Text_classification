# Custom Raw BERT Sentiment Analysis on IMDB Movie Reviews

## Project Objective
Build a complete NLP text classification system using a raw pretrained **BERT backbone** (`BertModel.from_pretrained('bert-base-uncased')`) with a custom **PyTorch Classification Head** (`Dropout` + `Linear` FC layer) to classify IMDB movie reviews into **Positive** or **Negative** sentiments.

### Pipeline Overview
```text
IMDB Dataset (50,000 reviews)
      ↓
Text Preprocessing & Cleaning
      ↓
Exploratory Data Analysis (EDA)
      ↓
Train / Validation / Test Split
      ↓
BERT Tokenization (input_ids & attention_masks)
      ↓
Raw BERT Backbone (BertModel) + Custom Head (Dropout + Linear FC Layer)
      ↓
Explicit CrossEntropyLoss Training & Early Stopping
      ↓
Training & Validation Curve Plots
      ↓
Final Test Set Evaluation & Metrics (Accuracy, Precision, Recall, F1, Confusion Matrix)
```


---
## Step 1: Import Dependencies & Setup System
We import scientific tools (`numpy`, `pandas`, `matplotlib`, `seaborn`), deep learning modules (`torch`, `torch.nn`), Hugging Face `BertModel` & `AutoTokenizer`, and `scikit-learn` metrics.

````python
import re
import time
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from torch.optim import AdamW
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    confusion_matrix, classification_report
)
from datasets import load_dataset
from transformers import AutoTokenizer, BertModel, get_linear_schedule_with_warmup
from tqdm import tqdm

# Set seeds for exact reproducibility
RANDOM_SEED = 42
torch.manual_seed(RANDOM_SEED)
np.random.seed(RANDOM_SEED)

# Auto-detect compute device (Apple Silicon MPS / CUDA GPU / CPU)
def get_device():
    if torch.backends.mps.is_available():
        return torch.device('mps')
    elif torch.cuda.is_available():
        return torch.device('cuda')
    else:
        return torch.device('cpu')

DEVICE = get_device()
print(f"Using compute device: {DEVICE}")
````

---
## Step 2: Central Configuration & Modular Parameters
Hyperparameters for training, sampling, max sequence length, and model settings.

````python
# Model settings
BERT_MODEL_NAME = 'bert-base-uncased'  # Pretrained checkpoint
MAX_LEN = 256                          # Maximum sequence length for BERT
NUM_CLASSES = 2                        # Binary classification (0: Neg, 1: Pos)
CLASS_NAMES = ['Negative', 'Positive']

# Dataset & Sampling Configuration
SAMPLE_MODE = True     # Set to False to run on the full 50,000 dataset!
SAMPLE_SIZE = 2000     # Subset size when SAMPLE_MODE is True

# Hyperparameters
TEST_SIZE = 0.20       # 20% reserved for testing
VAL_SIZE = 0.15        # 15% of training set reserved for validation
BATCH_SIZE = 16
EPOCHS = 4
LEARNING_RATE = 2e-5
PATIENCE = 2           # Early stopping patience
MODEL_SAVE_PATH = 'best_bert_sentiment_model.pt'

print(f"Configuration loaded | Model: {BERT_MODEL_NAME} | Sample Mode: {SAMPLE_MODE}")
````

---
## Step 3: Load Dataset & Exploratory Data Analysis (EDA)
We load the IMDB dataset, clean HTML artifacts (`<br />`), and compute class distribution.

````python
def clean_text(text: str) -> str:
    if not isinstance(text, str):
        return ""
    text = re.sub(r"<br\s*/?>", " ", text)
    text = re.sub(r"<[^>]+>", "", text)
    text = re.sub(r"\s+", " ", text).strip()
    return text

print("Loading IMDB dataset...")
raw_ds = load_dataset("imdb")
full_df = pd.concat([pd.DataFrame(raw_ds['train']), pd.DataFrame(raw_ds['test'])], ignore_index=True)
full_df['text'] = full_df['text'].apply(clean_text)

if SAMPLE_MODE and SAMPLE_SIZE < len(full_df):
    df = full_df.groupby('label', group_keys=False).apply(
        lambda x: x.sample(n=SAMPLE_SIZE // 2, random_state=RANDOM_SEED)
    ).reset_index(drop=True)
    print(f"Sample Mode Active: Sampled {len(df):,} reviews.")
else:
    df = full_df
    print(f"Full Dataset Active: {len(df):,} reviews.")

# Exploratory Data Analysis
pos_count = (df['label'] == 1).sum()
neg_count = (df['label'] == 0).sum()

print("\n" + "=" * 50)
print("         EXPLORATORY DATA ANALYSIS (EDA)")
print("=" * 50)
print(f"Total Reviews Analyzed      : {len(df):,}")
print(f"Positive Reviews (Label 1)  : {pos_count:,} ({pos_count/len(df)*100:.2f}%)")
print(f"Negative Reviews (Label 0)  : {neg_count:,} ({neg_count/len(df)*100:.2f}%)")
print("=" * 50)
````

---
## Step 4: Train / Validation / Test Splitting
Stratified dataset splits (65% Train / 15% Val / 20% Test).

````python
train_val_df, test_df = train_test_split(
    df, test_size=TEST_SIZE, random_state=RANDOM_SEED, stratify=df['label']
)

adjusted_val_size = VAL_SIZE / (1.0 - TEST_SIZE)
train_df, val_df = train_test_split(
    train_val_df, test_size=adjusted_val_size, random_state=RANDOM_SEED, stratify=train_val_df['label']
)

print(f"Dataset Splits:")
print(f"  Training Set   : {len(train_df):,} samples")
print(f"  Validation Set : {len(val_df):,} samples")
print(f"  Test Set       : {len(test_df):,} samples")
````

---
## Step 5: Tokenization & PyTorch DataLoaders
BERT Tokenizer encodes text into `input_ids` and `attention_mask` tensors.

````python
tokenizer = AutoTokenizer.from_pretrained(BERT_MODEL_NAME)

class IMDBDataset(Dataset):
    def __init__(self, texts, labels, tokenizer, max_len=MAX_LEN):
        self.texts = list(texts)
        self.labels = list(labels)
        self.tokenizer = tokenizer
        self.max_len = max_len

    def __len__(self):
        return len(self.texts)

    def __getitem__(self, idx):
        encoding = self.tokenizer(
            str(self.texts[idx]),
            add_special_tokens=True,
            max_length=self.max_len,
            padding='max_length',
            truncation=True,
            return_attention_mask=True,
            return_tensors='pt'
        )
        return {
            'input_ids': encoding['input_ids'].flatten(),
            'attention_mask': encoding['attention_mask'].flatten(),
            'labels': torch.tensor(self.labels[idx], dtype=torch.long)
        }

train_loader = DataLoader(IMDBDataset(train_df['text'], train_df['label'], tokenizer), batch_size=BATCH_SIZE, shuffle=True)
val_loader = DataLoader(IMDBDataset(val_df['text'], val_df['label'], tokenizer), batch_size=BATCH_SIZE, shuffle=False)
test_loader = DataLoader(IMDBDataset(test_df['text'], test_df['label'], tokenizer), batch_size=BATCH_SIZE, shuffle=False)

print("PyTorch DataLoaders successfully initialized!")
````

---
## Step 6: Define Custom Classifier with Raw BERT Backbone
We load raw `BertModel.from_pretrained('bert-base-uncased')` as the feature extractor backbone, and attach a custom PyTorch classification head (`Dropout(0.3)` + `nn.Linear(768, 2)`).

````python
class CustomBertClassifier(nn.Module):
    def __init__(self, model_name=BERT_MODEL_NAME, num_classes=NUM_CLASSES, dropout_rate=0.3):
        super(CustomBertClassifier, self).__init__()
        # Raw Pretrained BERT Backbone
        self.bert = BertModel.from_pretrained(model_name)
        # Custom Classification Head Layers
        self.dropout = nn.Dropout(p=dropout_rate)
        self.fc = nn.Linear(self.bert.config.hidden_size, num_classes)  # 768 -> 2

    def forward(self, input_ids, attention_mask):
        outputs = self.bert(input_ids=input_ids, attention_mask=attention_mask)
        pooled_output = outputs.pooler_output  # Extract 768-dim vector of [CLS] token
        x = self.dropout(pooled_output)
        logits = self.fc(x)  # Fully Connected Linear layer
        return logits

model = CustomBertClassifier(model_name=BERT_MODEL_NAME, num_classes=NUM_CLASSES)
model.to(DEVICE)
print(f"Custom Raw BERT Classifier loaded and transferred to device: {DEVICE}")
````

---
## Step 7: Model Fine-Tuning with CrossEntropyLoss & Early Stopping
We compute loss explicitly using `nn.CrossEntropyLoss()`, optimize with `AdamW`, and apply Early Stopping.

````python
criterion = nn.CrossEntropyLoss()
optimizer = AdamW(model.parameters(), lr=LEARNING_RATE, weight_decay=0.01)
total_steps = len(train_loader) * EPOCHS
scheduler = get_linear_schedule_with_warmup(optimizer, num_warmup_steps=int(total_steps * 0.1), num_training_steps=total_steps)

best_val_loss = float('inf')
patience_counter = 0
history = {'train_loss': [], 'train_acc': [], 'val_loss': [], 'val_acc': []}

print("Starting Custom BERT Training Loop...")
for epoch in range(1, EPOCHS + 1):
    # Training Epoch
    model.train()
    train_loss, train_correct, train_total = 0.0, 0, 0
    for batch in tqdm(train_loader, desc=f"Epoch {epoch}/{EPOCHS} [Train]"):
        input_ids = batch['input_ids'].to(DEVICE)
        attention_mask = batch['attention_mask'].to(DEVICE)
        labels = batch['labels'].to(DEVICE)

        optimizer.zero_grad()
        logits = model(input_ids=input_ids, attention_mask=attention_mask)
        loss = criterion(logits, labels)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
        optimizer.step()
        scheduler.step()

        train_loss += loss.item()
        preds = torch.argmax(logits, dim=1)
        train_correct += (preds == labels).sum().item()
        train_total += labels.size(0)

    epoch_train_loss = train_loss / len(train_loader)
    epoch_train_acc = train_correct / train_total

    # Validation Epoch
    model.eval()
    val_loss, val_correct, val_total = 0.0, 0, 0
    with torch.no_grad():
        for batch in val_loader:
            input_ids = batch['input_ids'].to(DEVICE)
            attention_mask = batch['attention_mask'].to(DEVICE)
            labels = batch['labels'].to(DEVICE)

            logits = model(input_ids=input_ids, attention_mask=attention_mask)
            loss = criterion(logits, labels)
            val_loss += loss.item()
            preds = torch.argmax(logits, dim=1)
            val_correct += (preds == labels).sum().item()
            val_total += labels.size(0)

    epoch_val_loss = val_loss / len(val_loader)
    epoch_val_acc = val_correct / val_total

    history['train_loss'].append(epoch_train_loss)
    history['train_acc'].append(epoch_train_acc)
    history['val_loss'].append(epoch_val_loss)
    history['val_acc'].append(epoch_val_acc)

    print(f"Epoch {epoch} Results -> Train Loss: {epoch_train_loss:.4f}, Train Acc: {epoch_train_acc*100:.2f}% | Val Loss: {epoch_val_loss:.4f}, Val Acc: {epoch_val_acc*100:.2f}%")

    # Early Stopping Check
    if epoch_val_loss < best_val_loss:
        print(f"  --> Val loss improved ({best_val_loss:.4f} -> {epoch_val_loss:.4f}). Saving model state...")
        best_val_loss = epoch_val_loss
        torch.save(model.state_dict(), MODEL_SAVE_PATH)
        patience_counter = 0
    else:
        patience_counter += 1
        print(f"  --> Val loss did not improve. Patience counter: {patience_counter}/{PATIENCE}")
        if patience_counter >= PATIENCE:
            print("[EARLY STOPPING] Validation performance plateaued. Stopping training.")
            break

# Restore best model state
model.load_state_dict(torch.load(MODEL_SAVE_PATH, map_location=DEVICE))
print("Restored best checkpoint successfully!")
````

---
## Step 8: Plot Training & Validation Curves
We plot Training Loss vs Validation Loss and Training Accuracy vs Validation Accuracy over epochs.

````python
epochs_range = range(1, len(history['train_loss']) + 1)
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].plot(epochs_range, history['train_loss'], 'b-o', label='Training Loss', linewidth=2)
axes[0].plot(epochs_range, history['val_loss'], 'r-s', label='Validation Loss', linewidth=2)
axes[0].set_title('Training vs Validation Loss', fontweight='bold')
axes[0].set_xlabel('Epochs')
axes[0].set_ylabel('Loss')
axes[0].legend()
axes[0].grid(True)

axes[1].plot(epochs_range, [acc * 100 for acc in history['train_acc']], 'b-o', label='Training Accuracy', linewidth=2)
axes[1].plot(epochs_range, [acc * 100 for acc in history['val_acc']], 'r-s', label='Validation Accuracy', linewidth=2)
axes[1].set_title('Training vs Validation Accuracy (%)', fontweight='bold')
axes[1].set_xlabel('Epochs')
axes[1].set_ylabel('Accuracy (%)')
axes[1].legend()
axes[1].grid(True)

plt.tight_layout()
plt.show()
````

---
## Step 9: Final Test Set Evaluation & Confusion Matrix
We evaluate the best model on the unseen Test Set and calculate metrics and confusion matrix.

````python
model.eval()
y_true, y_pred = [], []
with torch.no_grad():
    for batch in test_loader:
        input_ids = batch['input_ids'].to(DEVICE)
        attention_mask = batch['attention_mask'].to(DEVICE)
        labels = batch['labels'].to(DEVICE)

        logits = model(input_ids=input_ids, attention_mask=attention_mask)
        preds = torch.argmax(logits, dim=1)
        y_true.extend(labels.cpu().numpy())
        y_pred.extend(preds.cpu().numpy())

y_true, y_pred = np.array(y_true), np.array(y_pred)

acc = accuracy_score(y_true, y_pred)
prec = precision_score(y_true, y_pred, average='binary')
rec = recall_score(y_true, y_pred, average='binary')
f1 = f1_score(y_true, y_pred, average='binary')

cm = confusion_matrix(y_true, y_pred)
report = classification_report(y_true, y_pred, target_names=CLASS_NAMES, digits=4)

# Confusion Matrix Plot
plt.figure(figsize=(6, 5))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=CLASS_NAMES, yticklabels=CLASS_NAMES, annot_kws={"size": 14, "weight": "bold"})
plt.title('IMDB Test Confusion Matrix', fontweight='bold')
plt.xlabel('Predicted Sentiment')
plt.ylabel('Actual Sentiment')
plt.show()
````

---
## Step 10: Final Results Summary & Fitting Analysis

````python
print("=" * 60)
print("             FINAL MODEL EVALUATION SUMMARY")
print("=" * 60)
print(f"Model          : Custom Raw BERT ({BERT_MODEL_NAME})")
print(f"Dataset        : IMDB Movie Reviews ({'Subset Sample: ' + str(SAMPLE_SIZE) if SAMPLE_MODE else 'Complete 50,000 Dataset'})")
print(f"Task           : Binary Sentiment Classification")
print(f"Test Accuracy  : {acc * 100:.2f}%")
print(f"Precision      : {prec:.4f}")
print(f"Recall         : {rec:.4f}")
print(f"F1-Score       : {f1:.4f}")
print("=" * 60)

print("\n--- Classification Report ---")
print(report)

final_tr_loss, final_va_loss = history['train_loss'][-1], history['val_loss'][-1]
print("\n--- Diagnostic Fitting Analysis ---")
print(f"Final Training Loss: {final_tr_loss:.4f} | Final Validation Loss: {final_va_loss:.4f}")
if final_va_loss - final_tr_loss > 0.15:
    print("Diagnosis: Slight Overfitting observed. The model performs slightly better on training data than validation data.")
elif final_tr_loss > 0.5:
    print("Diagnosis: Underfitting observed. Additional epochs or hyperparameter tuning recommended.")
else:
    print("Diagnosis: Optimal Fitting achieved! Training and validation curves converged nicely.")
````
